# Linear Algebra Review

## 1. Matrix

A matrix \(A\) with \(m\) rows and \(n\) columns:

$$
\boxed{A\in\mathbb{R}^{m\times n}}
$$

Entry notation:

$$
\boxed{A_{ij}=\text{entry in row }i,\text{ column }j}
$$

**First index = row, second index = column.**

---

## 2. Transpose

The transpose switches rows and columns:

$$
\boxed{A\in\mathbb{R}^{m\times n}
\Rightarrow
A^T\in\mathbb{R}^{n\times m}}
$$

Example:

$$
A=
\begin{bmatrix}
1&2&3\\
4&5&6
\end{bmatrix}
\quad\Rightarrow\quad
A^T=
\begin{bmatrix}
1&4\\
2&5\\
3&6
\end{bmatrix}
$$

Important:

$$
\boxed{(A^T)^T=A}
$$

$$
\boxed{(AB)^T=B^TA^T}
$$

$$
\boxed{(A+B)^T=A^T+B^T}
$$

 When transposing a product, **reverse the order**.

---

## 3. Matrix Multiplication

If:

$$
A\in\mathbb{R}^{m\times n},
\qquad
B\in\mathbb{R}^{n\times p}
$$

then:

$$
\boxed{
(m\times n)(n\times p)=m\times p
}
$$

### Rule

$$
\boxed{\text{Inside dimensions must match}}
$$

$$
\boxed{\text{Result dimensions = outside dimensions}}
$$

Example:

$$
(2\times3)(3\times4)=2\times4
$$

Each entry is:

$$
\boxed{
(AB)_{ij}
=
\text{Row }i\text{ of }A
\cdot
\text{Column }j\text{ of }B
}
$$

Formally:

$$
(AB)_{ij}
=
\sum_{k=1}^{n}A_{ik}B_{kj}
$$

 Matrix multiplication is generally **not commutative**:

$$
\boxed{AB\ne BA}
$$

---

## 4. Matrix Multiplication Properties

Associative:

$$
\boxed{(AB)C=A(BC)}
$$

Distributive:

$$
\boxed{A(B+C)=AB+AC}
$$

 You can change parentheses, but **never change the order**.

---

## 5. Identity Matrix

The identity matrix has:

* \(1\)'s on the diagonal
* \(0\)'s everywhere else

$$
I_3=
\begin{bmatrix}
1&0&0\\
0&1&0\\
0&0&1
\end{bmatrix}
$$

It behaves like the number \(1\):

$$
\boxed{AI=A}
\qquad
\boxed{IA=A}
$$


# Vector Algebra

## 1. Vectors

An \(n\)-dimensional vector is a column vector:

$$
\boxed{
\mathbf{x}\in\mathbb{R}^n
=
\begin{bmatrix}
x_1\\
x_2\\
\vdots\\
x_n
\end{bmatrix}
}
$$

$$
\boxed{x_i=\text{\(i\)-th entry of }\mathbf{x}}
$$

---

## 2. Inner Product (Dot Product)

For \(\mathbf{x},\mathbf{y}\in\mathbb{R}^n\):

$$
\boxed{
\mathbf{x}^T\mathbf{y}
=
\sum_{i=1}^{n}x_i y_i
}
$$

Dimension:

$$
(1\times n)(n\times1)=1\times1
$$

Result: **scalar**

$$
\boxed{\mathbf{x}^T\mathbf{y}=\mathbf{y}^T\mathbf{x}}
$$

---

## 3. Outer Product

$$
\boxed{\mathbf{x}\mathbf{y}^T}
$$

Dimension:

$$
(n\times1)(1\times n)=n\times n
$$

Result: **matrix**

Each entry:

$$
\boxed{
(\mathbf{x}\mathbf{y}^T)_{ij}=x_i y_j
}
$$

**Remember:**

$$
\boxed{\mathbf{x}^T\mathbf{y}\rightarrow\text{scalar}}
$$

$$
\boxed{\mathbf{x}\mathbf{y}^T\rightarrow\text{matrix}}
$$

---

## 4. Matrix × Vector

If:

$$
A\in\mathbb{R}^{m\times n},
\qquad
\mathbf{x}\in\mathbb{R}^{n}
$$

then:

$$
\boxed{
A\mathbf{x}\in\mathbb{R}^{m}
}
$$

$$
(m\times n)(n\times1)=m\times1
$$

Each entry:

$$
\boxed{
(A\mathbf{x})_i
=
\sum_{j=1}^{n}A_{ij}x_j
}
$$

---

# Eigenvalues and Eigenvectors

For a square matrix \(A\):

$$
\boxed{
A\mathbf{v}=\lambda\mathbf{v}
}
$$

* \($\mathbf{v}$\): eigenvector
* \($\lambda$\): eigenvalue

**Intuition:** \(A\) transforms \(\mathbf{v}\) without changing its direction, only its size/sign.

---

## Symmetric Matrix

$$
\boxed{A=A^T}
$$

For symmetric matrices:

* Eigenvalues are real.
* Eigenvectors are real.
* Eigenvectors for different eigenvalues are orthogonal:

$$
\boxed{
\mathbf{v}_i^T\mathbf{v}_j=0
}
$$
