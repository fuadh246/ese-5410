$$ \hat{y_\text{Graduation Rate}} = \hat{\beta_0}+\hat{\beta_1} X_{\text{Top25perc}}$$

We now cover some important statistical terms such as $R^2$ and Residual Standard Error.

$$ \text{RSS (Residual Sum of Squares)} = \sum_{i = 1}^{N}(y_i-\hat{y_i})^2$$

$$\text{RSE (Residual Standard Error)} = \sqrt{\frac{RSS}{DF}} = \sqrt{\frac{RSS}{n-p-1}} = \sqrt{\frac{RSS}{n-2}}$$

Where DF is the degrees of freedom

$$\text{TSS (Total Sum of Squares)} = \sum_{i = 1}^{N}(y_i-\mu)^2$$

Where $\mu$ is the average of $y$.

Now the famous $R^2$ is,

$$R^2 = \frac{TSS-RSS}{TSS}$$

```python
#Programming the above formulas
RSS = np.sum(fit1.resid**2) #Residuals is y_i - \hat{y_i}
RSE = np.sqrt(RSS/fit1.df_resid)
print("RSE:",RSE)
TSS = np.sum((data2['Grad_Rate']-np.average(data2['Grad_Rate']))**2)
Rsquared = (TSS-RSS)/TSS
print("R^2:",Rsquared)
```

```python
#The above code is shown for illustrative reasons, we can much easily access R^2
print("R^2:",fit1.rsquared)
```

```python
#Accessing RSE is also relatively easy
print("RSE:",np.sqrt(fit1.scale))
```

Residual Standard Error (RSE) is important because this is the estimate of $\sigma$ when we assumed the error term $\epsilon$ to be normally distributed with $N(0,\sigma)$

### Confidence Intervals

* $\hat{\beta}$ are the mean estimates

* Every estimate needs a confidence interval (Obtained from its mean and standard error)

* Confidence Interval
 $$\mu \pm (z\times \text{Standard Error})$$

### 95% Confidence intervals
$\beta_0$:

$$42.3651 \pm (1.96 \times 1.621) = [39.18,45.54]$$

$\beta_1$:

$$0.4140 \pm (1.96 \times 0.027) = [0.36,0.46]$$

We may interpret this as, with a 95% probability the true value of $\beta_1$ is contained within the interval $[0.36,0.46]$.

### Hypothesis Testing and p-values

* Is $\hat{\beta_1}$ statistically significant?

### Problem Formulation

* Null Hypothesis 
    * $H_0:$ No relationship between $X$ and $Y$
    * $H_0: \hat{\beta_1} = 0$

 * Alternative Hypothesis
     * $H_a:$ There is some relationship between $X$ and $Y$
     * $H_a: \hat{\beta_1} \neq 0$

### Interpretation

* A small **p-value**:
    * We reject the null hypothesis at a **particular significance level**
    * There is a relationship between Top25perc and Grad_Rate at a **particular significance level**
* A large **p-value**
    * We fail to reject the null hypothesis at a **particular significance level**
    * There is no association between Top25perc and Grad_Rate at a **particular significance level**
    
**We never accept the null hypothesis, we only fail to reject it**

```python
#P-values:
## The summary tables above show it as 0 for both coefficients, however this is impossible. The numbers are just small
fit0.pvalues
```

The interpretation of the p-value is, at a 0.05 level, we reject the null hypothesis. Hence, at a 0.05 level we believe there to be a significant relationship between Top25perc and Grad_rate because $0.05>1.87\times 10^{-45}$

Testing the null hypothesis for 3.($H_0: \beta_1=\beta_2 = 0$) is not as straight forward.

## Boston Dataset

The sklearn.datasets library contains the Boston dataset, which records the medv (median house value) for 506 neighborhoods around Boston. We will seek to predict MEDV using 13 predictors such as RM (average number of rooms per house), AGE (average age of house), and LSTAT (percent of households with low socioeconomic status). 

Here, Boston is a dictionary of the data with keys 'target', 'data' and 'DESCR' (documentation: https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_boston.html).

```python
from sklearn.datasets import load_boston
boston_dataset = load_boston()
Boston = pd.DataFrame(data = boston_dataset.data, columns = boston_dataset.feature_names)
Boston['MEDV'] = pd.Series(boston_dataset.target)
Boston.head()
```

### Multiple Linear Regression

In order to fit a multiple linear regression model using least squares, we again use the sm.OLS() function. The syntax is similar, but X is now an $n$x$p$ matrix of the features. The summary() function now outputs the regression coefficients for all the predictors.

```python
model = smf.ols('MEDV ~ LSTAT + AGE', data=Boston)
res = model.fit()
print(res.summary())
```

The Boston data set contains 13 variables, and so it would be cumbersome to have to type all of these in order to perform a regression using all of the predictors. Instead, we can use the following list comprehension that appends the names of the columns separated by +:

```python
model = smf.ols('MEDV ~' + ' + '.join([i for i in Boston.columns[:-1]]), data=Boston)
res = model.fit()
print(res.summary())
```

We can use model.fit().rsquared to get the $R^2$, and model.fit().mse_resid gives us the residual mean squared error. The variance_inflation_factor() function can be used to calculate the variance inflation factors. Most VIF's are low to moderate for this data.

```python
X = Boston.loc[:, Boston.columns != 'MEDV']

vif = pd.DataFrame([variance_inflation_factor(X.values, i) for i in range(X.shape[1])])
vif = pd.concat([pd.DataFrame(X.columns), vif], axis=1)
vif.columns = ['Feature','VIF']
vif
```

What if we would like to perform a regression using all of the variables but one? For example, in the above regression output, age has a high p-value. So we may wish to run a regression excluding this predictor. The following syntax results in a regression using all the predictors except age. 

```python
model = smf.ols('MEDV ~' + ' + '.join([i for i in X.columns]) + '-AGE', data=Boston)
res = model.fit()
print(res.summary())
```

### Interaction Terms

It is easy to include interaction terms in a linear model using the smf.ols() function. The syntax lstat:black tells Python to include an interaction term between lstat and black. The syntax lstat * age simultaneously includes lstat, age and the interaction term lstat x age as predictors; it is shorthand for lstat + age + lstat:age. 

```python
model = smf.ols('MEDV ~ LSTAT*AGE', data=Boston)
res = model.fit()
print(res.summary())
```

### Non-linear Transformations of the Predictors

The ols() function can also accomodate non-linear transformations of the predictors. For instance, given a predictor X, we can create a predictor X$^2$ using vectorized functions. We now perform a regression of medv onto lstat and lstat$^2$.

```python
model = smf.ols('MEDV ~ LSTAT + np.square(LSTAT)', data=Boston)
res2 = model.fit()
print(res2.summary())
```

The near-zero p-value associated with the quadratic term suggests that it leads to an improved model. We use the anova() function to further quantify the extent to which the quadratic fit is superior to the linear fit. 

```python
res1 = smf.ols('MEDV ~ LSTAT', data=Boston).fit()
aov_table = sm.stats.anova_lm(res1, res2)
aov_table
```

Here Res 1 represents the linear submodel containing only 1 predictor, lstat, while Res 2 corresponds to the larger quadratic model that has two predictors, lstat and lstat$^2$. The anova_lm() function performs a hypothesis test comparing the two models. The null hypothesis is that the two models fit the data equally well, and the alternative hypothesis is that the full model is superior. Here the F-statistic is 135 and the associated p-value is virtually zero. This provides very clear evidence that the model containing the predictors lstat and lstat$^2$ is far superior to the model that only contains the predictor lstat. This is not surprising, since earlier we saw evidence for non-linearity in the relationship between medv and lstat. If we type:

In order to create a cubic fit, we can include a predictor in the form np.power(X, 3). However, this approach can start to get cumbersome for higher order polynomials. A better approach involves using the vander() function to transform the variables. Alternatively, you could use numpy's polyfit() function.

```python
model5 = smf.ols('MEDV ~ np.vander(Boston.LSTAT, 6)[:,:5]', data=Boston).fit()
print(model5.summary())
```

This suggests that including additional polynomial terms, up to the fifth order, leads to an improvement in the model fit! However, further investigation of the data reveals that no polynomial terms beyond the fifth order have significant p-values in a regression fit.
