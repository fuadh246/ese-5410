### Rename columns

Renaming the column names of a dataframe is a good habit. It's not only for the later convenient use, but python is picky for the format of column names. So we can use [rename](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.rename.html) to replace the space and special symbols with underlines on column names.



```
# Rename some columns
df = df.rename(columns = {'Sale Price/bid price':'Price',
                              'Sheriff Cost': 'Sheriff_Cost',
                              ' Avg Walk&Transit score  ':'Avg_Transit_Score',
                              ' Violent Crime Rate ':'Violent_Crime_Rate',
                              ' School Score  ':'School_Score',
                              'finished \n(SqFt)':'SqFt',
                              ' bathrooms ':'Bathrooms',
                              ' bedrooms ':'Bedrooms', 
                              'Average comps':'Avg_Comps'})
```

### Drop Missing Values

As data comes in many shapes and forms, pandas aims to be flexible with regard to handling missing data. What we do here is to use [dropna](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.dropna.html) to drop the row if all values are NA in that row.

```
# Drop the row if all values are NA in that row
df.dropna(how = 'all',inplace = True)
df.reset_index(inplace = True,drop = True)
```

### Apply and Lambda Usage in Pandas

After observing the "yearBuilt" column, we noticed that the format of the values in this column is like "2020". Since we definitely want to take this column into consideration for our model, then we can convert this column to the number of years built. (2012 to be 2021-2012 = 9). The function we are using is [apply](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.apply.html). You can also refer to this [link](https://towardsdatascience.com/apply-and-lambda-usage-in-pandas-b13a1ea037f7) to know more about the apply and lambda usage in pandas.

```
# Convert the column yearBuilt to year. e.g. 2012 converted to 2021-2012 = 9 
df['years'] = df['yearBuilt'].apply(lambda x:2021-x)
df.drop('yearBuilt',axis=1,inplace = True)
```

For "Price" column, we need to remove special symbols like "$", comma, and space and convert the type from string to float. So does column "Avg_Comps"

```
# Convert price column from string type to float type
df['Price'] = df['Price'].apply(lambda x:x.replace('$',''))
df['Price'] = df['Price'].apply(lambda x:x.replace(' ',''))
df['Price'] = df['Price'].apply(lambda x:float(x.replace(',','')))
```

```
# Average comparables column
df['Avg_Comps'] = df['Avg_Comps'].apply(lambda x:float(x.replace(',','')))
```

### Convert a Categorical Column into Dummy Columns 

For column "PropType", we use [get_dummies](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.get_dummies.html) to convert this categorical variable into dummy/indicator variables. Then use [concat](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.concat.html) to concatenate pandas objects along a particular axis with optional set logic along the other axes.

```
# One hot encoding
df = pd.concat([df,pd.get_dummies(df['PropType'], prefix='Type')],axis = 1)
df.drop('PropType',axis = 1,inplace = True)
```

### Imputation

Imputation is not necessarily in this course. When should we use imputation to deal with missing values?

```
# Regression Imputation on Bathroom column
df['Bathrooms'] = df['Bathrooms'].apply(lambda x:eval(x.replace(' -   ','None')))

missing_bathroom_index = df[df['Bathrooms'].isna()].index

known_bathrooms = df[~df['Bathrooms'].isna()]
unknown_bathrooms = df[df['Bathrooms'].isna()]

known_bathrooms_x = known_bathrooms[['SqFt','Type_Condominium', 'Type_MultiFamily2To4', 
                                     'Type_SingleFamily','Type_Townhouse', 'years']]
known_bathrooms_y = known_bathrooms['Bathrooms']

unknown_bathrooms_x = unknown_bathrooms[['SqFt','Type_Condominium', 'Type_MultiFamily2To4', 
                                     'Type_SingleFamily','Type_Townhouse', 'years']]

imputation_model = LinearRegression()
imputation_model.fit(known_bathrooms_x,known_bathrooms_y)
for i in missing_bathroom_index:
  df.at[i,'Bathrooms'] = round(imputation_model.predict(unknown_bathrooms_x.loc[[i]])[0])
```

```
# Regression Imputation on Bedroom column
df['Bedrooms'] = df['Bedrooms'].apply(lambda x:eval(x.replace(' -   ','None')))

missing_bedroom_index = df[df['Bedrooms'].isna()].index

known_bedrooms = df[~df['Bedrooms'].isna()]
unknown_bedrooms = df[df['Bedrooms'].isna()]

known_bedrooms_x = known_bedrooms[['SqFt','Type_Condominium', 'Type_MultiFamily2To4', 
                                     'Type_SingleFamily','Type_Townhouse', 'years']]
known_bedrooms_y = known_bedrooms['Bedrooms']

unknown_bedrooms_x = unknown_bedrooms[['SqFt','Type_Condominium', 'Type_MultiFamily2To4', 
                                     'Type_SingleFamily','Type_Townhouse', 'years']]

imputation_model = LinearRegression()
imputation_model.fit(known_bedrooms_x,known_bedrooms_y)
for i in missing_bedroom_index:
  df.at[i,'Bedrooms'] = round(imputation_model.predict(unknown_bedrooms_x.loc[[i]])[0])
```

But don't worry! We can parse all of the school information from the website and then mark them on the map as well.

```
import requests
from lxml import html

w = requests.get("https://en.wikipedia.org/wiki/List_of_schools_of_the_School_District_of_Philadelphia")
dom_tree = html.fromstring(w.content)
```

```
school1 = pd.Series(dom_tree.xpath('//div[contains(@class,"mw-parser")]//ul/li/text()')).to_frame('School')
school2 = pd.Series(dom_tree.xpath('//div[contains(@class,"mw-parser")]//ul/li/a/text()')).to_frame('School')
school = pd.concat([school1,school2],axis = 0)
```

The data parsed from wikipedia would also contain some useless information and symbols that we don't need. So we also do some data cleaning here.

```
school['School'] = school['School'].apply(lambda x:x.replace("\n",""))
school['School'] = school['School'].apply(lambda x:x.replace(",",""))
school['School'] = school['School'].apply(lambda x:str(0) if "(" in x else x)
school['School'] = school['School'].apply(lambda x:x if "School" in x else None)
school = school[~school['School'].isna()]
school.reset_index(inplace = True,drop = True)
```

```
school['Location'] = school['School'].apply(lambda x:geocodeN(x))
school['Longitude'] = school['Location'].apply(lambda x:x[0])
school['Latitude'] = school['Location'].apply(lambda x:x[1])
school = school[school['Location'] != (0,0)]
```
