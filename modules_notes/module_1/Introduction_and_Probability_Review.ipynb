# Elements of Probability

## 1. Sample Space, Events, and Probability

Probability theory is built on three basic elements:

### Sample Space

The **sample space** $\Omega$ is the set of all possible outcomes of an experiment.

$$
\Omega = \{\text{all possible outcomes}\}
$$

For example, when tossing a six-sided die:

$$
\Omega = \{1,2,3,4,5,6\}
$$

---

### Events

An **event** $A$ is a subset of the sample space:

$$
A \subseteq \Omega
$$

The collection of events is denoted by $\mathcal{F}$:

$$
\mathcal{F} \subseteq 2^\Omega
$$

For a finite sample space, we can often take:

$$
\mathcal{F} = 2^\Omega
$$

where $2^\Omega$ is the set of **all subsets** of $\Omega$.

For the die example, some possible events are:

$$
A = \{2,4,6\} \qquad \text{(even)}
$$

$$
B = \{1,3,5\} \qquad \text{(odd)}
$$

$$
C = \{1,2,3\}
$$

---

### Probability Measure

A probability measure $P$ maps events to real numbers:

$$
P:\mathcal{F}\rightarrow[0,1]
$$

We interpret $P(A)$ as the probability that event $A$ occurs.

### The Three Axioms of Probability

**Axiom 1 — Non-negativity**

$$
P(A)\geq 0
$$

**Axiom 2 — Total probability**

$$
P(\Omega)=1
$$

**Axiom 3 — Countable additivity**

If $A_1,A_2,\ldots$ are mutually disjoint events, then

$$
P\left(\bigcup_{i=1}^{\infty}A_i\right)
=
\sum_{i=1}^{\infty}P(A_i)
$$

For two disjoint events:

$$
A\cap B=\varnothing
$$

we have

$$
P(A\cup B)=P(A)+P(B)
$$

---

# 2. Example: Rolling a Die

For a fair six-sided die:

$$
\Omega=\{1,2,3,4,5,6\}
$$

Since every outcome is equally likely:

$$
P(A)=\frac{|A|}{|\Omega|}
$$

where $|A|$ denotes the **cardinality** (number of elements) of a set.

### Example: Probability of an even number

Let

$$
A=\{2,4,6\}
$$

Then

$$
|A|=3
$$

and

$$
|\Omega|=6
$$

Therefore,

$$
P(A)=\frac{|A|}{|\Omega|}
=\frac{3}{6}
=\frac12
$$

---

# 3. Useful Properties of Probability

### Property 1 — Intersection

For any two events $A$ and $B$:

$$
P(A\cap B)\leq \min\{P(A),P(B)\}
$$

Intuitively, $A\cap B$ is contained inside both $A$ and $B$:

$$
A\cap B\subseteq A
$$

and

$$
A\cap B\subseteq B
$$

---

### Property 2 — Union Bound

For any two events:

$$
P(A\cup B)\leq P(A)+P(B)
$$

This is called the **union bound**.

If $A$ and $B$ are disjoint, then equality holds:

$$
P(A\cup B)=P(A)+P(B)
$$

---

### Property 3 — Partition of the Sample Space

If $A_1,\ldots,A_k$ form a **partition** of $\Omega$, then:

$$
\Omega=\bigcup_{i=1}^{k}A_i
$$

and the events are mutually disjoint:

$$
A_i\cap A_j=\varnothing
\qquad i\neq j
$$

Therefore:

$$
\sum_{i=1}^{k}P(A_i)=1
$$

---

# 4. Conditional Probability

The **conditional probability** of $A$ given $B$ is:

$$
P(A\mid B)
=
\frac{P(A\cap B)}{P(B)}
$$

where

$$
P(B)>0
$$

### Intuition

$P(A\mid B)$ means:

> What is the probability of $A$ **after knowing that $B$ occurred?**

The denominator $P(B)$ acts as a normalization factor.

---

## Conditional Probability from a Venn Diagram

The intersection $A\cap B$ represents the outcomes where **both $A$ and $B$ occur**.

Therefore:

$$
P(A\mid B)
=
\frac{\text{area}(A\cap B)}
{\text{area}(B)}
$$

Notice that we are effectively restricting our sample space to $B$.

---

# 5. Independence

Two events $A$ and $B$ are **independent** if:

$$
P(A\cap B)=P(A)P(B)
$$

This means that knowing whether $B$ occurred does **not change** the probability of $A$.

Equivalently, if $P(B)>0$:

$$
P(A\mid B)=P(A)
$$

### Proof

Start with the definition of conditional probability:

$$
P(A\mid B)
=
\frac{P(A\cap B)}{P(B)}
$$

If $A$ and $B$ are independent:

$$
P(A\cap B)=P(A)P(B)
$$

Therefore:

$$
P(A\mid B)
=
\frac{P(A)P(B)}{P(B)}
$$

Cancel $P(B)$:

$$
\boxed{P(A\mid B)=P(A)}
$$

### Intuition

If learning that $B$ occurred doesn't change the probability of $A$, then $A$ and $B$ are independent.

# 6. Random Variables

A **random variable** is a function that maps outcomes in the sample space to real numbers.

$$
X:\Omega\rightarrow\mathbb{R}
$$

For an outcome $\omega\in\Omega$:

$$
X(\omega)\in\mathbb{R}
$$

We usually use:

* $\Omega$ = sample space
* $\omega$ = one particular outcome
* $X$ = random variable
* $X(\omega)$ = value of the random variable for outcome $\omega$
* $x$ = a possible realization/value of $X$

So:

$$
\boxed{\omega \xrightarrow{\ X\ } X(\omega)}
$$

A random variable is therefore a **function**, not the outcome itself.

---

# 7. Example: Tossing Three Coins

Suppose we toss three coins simultaneously.

The sample space is:

$$
\Omega=
\{
HHH,HHT,HTH,HTT,
THH,THT,TTH,TTT
\}
$$

There are:

$$
|\Omega|=2^3=8
$$

possible outcomes.

---

## Defining a Random Variable

Let $X$ be the **number of heads**:

$$
X:\Omega\rightarrow\mathbb{R}
$$

We define:

$$
X(\omega)=\text{number of heads in }\omega
$$

For example:

$$
X(HHH)=3
$$

$$
X(HHT)=2
$$

$$
X(HTH)=2
$$

$$
X(HTT)=1
$$

$$
X(TTT)=0
$$

Therefore, the random variable maps:

$$
\begin{aligned}
HHH &\rightarrow 3\\
HHT &\rightarrow 2\\
HTH &\rightarrow 2\\
HTT &\rightarrow 1\\
THH &\rightarrow 2\\
THT &\rightarrow 1\\
TTH &\rightarrow 1\\
TTT &\rightarrow 0
\end{aligned}
$$

The important idea is:

$$
\boxed{\text{Outcome }\omega
\quad\xrightarrow{\text{random variable }X}\quad
\text{number }X(\omega)}
$$

For example:

$$
\omega=HHT
$$

and

$$
X(\omega)=2
$$

---

# 8. Key Distinction: Outcome vs. Random Variable

This distinction is extremely important.

### Outcome

An outcome is an element of the sample space:

$$
\omega\in\Omega
$$

Example:

$$
\omega=HHT
$$

### Random Variable

A random variable is a function:

$$
X:\Omega\rightarrow\mathbb{R}
$$

### Realization

The value obtained after applying $X$ to an outcome is:

$$
X(\omega)=x
$$

For example:

$$
HHT
\xrightarrow{X}
2
$$

So:

$$
\boxed{
\omega=HHT,\qquad X(\omega)=2
}
$$

---

# Summary

The basic hierarchy is:

$$
\boxed{
\Omega
\rightarrow
\mathcal{F}
\rightarrow
P
}
$$

where:

* $\Omega$ = sample space
* $\mathcal{F}$ = collection of events
* $P$ = probability measure

Then we introduce random variables:

$$
\boxed{X:\Omega\rightarrow\mathbb{R}}
$$

And conditional probability:

$$
\boxed{
P(A\mid B)=\frac{P(A\cap B)}{P(B)}
}
$$


And independence:

$$
\boxed{
P(A\cap B)=P(A)P(B)
}
$$

which, when $P(B)>0$, is equivalent to:

$$
\boxed{
P(A\mid B)=P(A)
}
$$


## Probability Distributions

A probability distribution is a function that describes the likelihood of obtaining the possible values of a random variable.

## 1. Discrete Probability Distributions aka Probability Mass Functions (PMFs)

A discrete probability distribution is a function that assigns probabilities to each possible value of a discrete random variable.

$$
p_x(x) = P(X = x)
$$

where $p_x(x)$ is the probability that the random variable $X$ takes the value $x$.

The function $p_x(x)$ must satisfy:

1. $p_x(x) \geq 0$ for all $x$
2. $\sum_{x} p_x(x) = 1$



```python
import seaborn as sns
import matplotlib.pyplot as plt

# Data
x = [0, 1, 2, 3]
p = [1/8, 3/8, 3/8, 1/8]

# Style
sns.set_theme(style="whitegrid")

# Plot
ax = sns.barplot(x=x, y=p)

# Labels
ax.set_xlabel("Number of Heads ($X$)")
ax.set_ylabel("$P(X=x)$")
ax.set_title("PMF: Number of Heads in 3 Coin Flips")

# Add probability labels
for i, probability in enumerate(p):
    ax.text(
        i,
        probability + 0.01,
        f"{probability:.3f}",
        ha="center"
    )

plt.ylim(0, 0.5)
plt.show()
```

### Interpretation

Each stem represents the probability of a particular value of $X$.

For example:

$$
P(X=2)=\frac{3}{8}
$$

meaning there is a $\frac{3}{8}$ probability of getting exactly 2 heads
when flipping 3 fair coins.

Because $X$ is discrete, the PMF assigns probability to individual values.
